# TWF', 'HDF', 'PWF','OSF','RNF' These columns describe specific failure events and are used to construct the overall machine failure label. Giving them to the model would reveal information about the outcome we're trying to predict.
This is called data leakage.
also we dont need the 'UDI','Product ID' columns 

```python
y= df["Machine failure"]
columns_to_drop =['Machine failure','TWF','HDF','PWF','OSF','RNF','UDI','Product ID']
x= df.drop(columns_to_drop,axis=1)
print(x.head())
print(y.head())
print(y.value_counts())
```

# our model use six input features : Type, Air Temperature, Process Temperature, Rotational speed, Torque, Tool wear
 

```python
x["Temperature difference"] = (x["Process temperature [K]"]-x["Air temperature [K]"])
x.head()
```

# we created a new feature Temperature difference .The difference between process and surrounding air temperature may help describe the machine's thermal operating conditions.    


```python
# categorical features
x.select_dtypes(include=["object"]).columns.tolist()
```

```python
# numerical features
x.select_dtypes(include=["number"]).columns.tolist()
```

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test=train_test_split(x,y,test_size=0.20,random_state=42,stratify=y)
 
```

```python
import os

os.makedirs("../data/processed", exist_ok=True)

x_train.to_csv("../data/processed/x_train.csv", index=False)
x_test.to_csv("../data/processed/x_test.csv", index=False)
y_train.to_csv("../data/processed/y_train.csv", index=False)
y_test.to_csv("../data/processed/y_test.csv", index=False)

print("Datasets saved successfully!")
```
